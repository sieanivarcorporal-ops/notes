```python
def rotate_word(word):
    return word[1: ] + word[0]

rotate_word("Python")
```

```python
def make_username(first_name, last_name):
    first = first_name.lower().replace(" ", "")
    last = last_name.lower().replace(" ", "")
    return first + "." + last

make_username("Ivar", "Corporal")
```

```python
def swap_bookends(items):
    first, *middle, last = items
    return [last] + middle + [first]

print(swap_bookends([1,2,3,4,5,6]))
```

```python
def reverse(text):
    reversed = text[-1: ] + text[2: : -1]
    return reversed

reverse("ivar")
```

```python
def make_email(first_name, last_name, domain):
    first = first_name[0].replace(" ","").lower()
    last = last_name.lower().replace(" ", "")
    domain = domain.lower()
    return first + "." + last + "@" + domain

make_email("Siean Ivar", "Corporal", "ust.edu.ph")
```

```python
def swap_middle_ends(content):
    first, second, *middle, second_last, last = content
    return [first] + [second_last] + middle + [second] + [last]

swap_middle_ends([1,2,3,4,5,6,7,8])
```
