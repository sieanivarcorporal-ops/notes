## Problem 1: Word Reversal Rotation — `rotate_and_reverse()`

Accepts a non-empty string. Move the **last** character to the front, then reverse the
remaining characters.

```
rotate_and_reverse("python") -> "nnohty"
rotate_and_reverse("logic")  -> "cgilo"
```

**Requirement:** Use slicing only (no loops).

````python
def rotate_and_reverse(text):
    last_char = text[-1]
    rest_reversed = text[:-1][::-1]
    return last_char + rest_reversed
````

**Test cases:**

````python
print(rotate_and_reverse("python"))  # expected: nnohty
print(rotate_and_reverse("logic"))   # expected: cgilo
print(rotate_and_reverse("A"))       # expected: A
````

## Problem 2: Email Builder — `make_email()`

Accepts `first_name`, `last_name`, and `domain`. Builds an email in the format
`f.lastname@domain`, where the first name is reduced to its first letter (lowercase,
no spaces) and the domain is lowercased.

```
make_email("Ada", "Lovelace", "UST.EDU.PH") -> "a.lovelace@ust.edu.ph"
make_email("Ana Maria", "De Leon", "gmail.com") -> "a.deleon@gmail.com"
```

````python
def make_email(first_name, last_name, domain):
    first_initial = first_name.strip().replace(" ", "")[0].lower()
    clean_last = last_name.lower().replace(" ", "")
    clean_domain = domain.lower()
    return f"{first_initial}.{clean_last}@{clean_domain}"
````

**Test cases:**

````python
print(make_email("Ada", "Lovelace", "UST.EDU.PH"))       # expected: a.lovelace@ust.edu.ph
print(make_email("Ana Maria", "De Leon", "gmail.com"))    # expected: a.deleon@gmail.com
print(make_email("Alan", "Turing", "Outlook.COM"))        # expected: a.turing@outlook.com
````

## Problem 3: Middle Swap — `swap_middle_ends()`

Accepts a list with at least 4 elements. Unpacks into `first`, `second`, `*middle`,
`second_last`, `last`. Returns a new list where `second` and `second_last` have
swapped positions; everything else stays the same.

```
swap_middle_ends([1, 2, 3, 4, 5, 6]) -> [1, 5, 3, 4, 2, 6]
```

**Requirement:** Use extended sequence unpacking.

````python
def swap_middle_ends(items):
    first, second, *middle, second_last, last = items
    return [first, second_last, *middle, second, last]
````

**Test cases:**

````python
print(swap_middle_ends([1, 2, 3, 4, 5, 6]))            # expected: [1, 5, 3, 4, 2, 6]
print(swap_middle_ends(["a", "b", "c", "d"]))          # expected: ['a', 'c', 'b', 'd']
print(swap_middle_ends([10, 20, 30, 40, 50]))          # expected: [10, 40, 30, 20, 50]
````

## Problem 4: Caesar-Style Shift — `shift_letters()`

Accepts a string and an integer `n`. Shifts every letter forward by `n` positions in the
alphabet (wrapping from z back to a), preserving case, and leaving non-letters unchanged.

```
shift_letters("abc", 2)    -> "cde"
shift_letters("XYZ", 3)    -> "ABC"
shift_letters("Code 1!", 1) -> "Dpef 1!"
```

**Requirement:** No explicit loops — use a comprehension and `ord`/`chr`.

````python
def shift_letters(text, n):
    def shift_char(c):
        if c.isupper():
            return chr((ord(c) - ord("A") + n) % 26 + ord("A"))
        elif c.islower():
            return chr((ord(c) - ord("a") + n) % 26 + ord("a"))
        else:
            return c
    return "".join(shift_char(c) for c in text)
````

**Test cases:**

````python
print(shift_letters("abc", 2))     # expected: cde
print(shift_letters("XYZ", 3))     # expected: ABC
print(shift_letters("Code 1!", 1)) # expected: Dpef 1!
````

## Problem 5: Palindrome Checker with Cleanup — `is_clean_palindrome()`

Accepts a string that may contain spaces, punctuation, and mixed case. Strips all
non-letter characters, lowercases everything, and checks whether the result reads the
same forwards and backwards.

```
is_clean_palindrome("A man, a plan, a canal: Panama") -> True
is_clean_palindrome("Was it a car or a cat I saw?")    -> True
is_clean_palindrome("Hello")                            -> False
```

**Requirement:** No explicit loops for the check — use `s == s[::-1]`.

````python
def is_clean_palindrome(text):
    cleaned = "".join(c.lower() for c in text if c.isalpha())
    return cleaned == cleaned[::-1]
````

**Test cases:**

````python
print(is_clean_palindrome("A man, a plan, a canal: Panama"))  # expected: True
print(is_clean_palindrome("Was it a car or a cat I saw?"))     # expected: True
print(is_clean_palindrome("Hello"))                             # expected: False
````

## Problem 6: Nested Unpacking Table — `pivot_row()`

Accepts a list of the form `[id, name, *scores, average]`, where `scores` is variable-length.
Unpacks it, recomputes the average from `scores` (rounded to 2 decimals), and returns a new
list `[id, name, *scores, recomputed_average]`.

```
pivot_row([1, "Ana", 90, 85, 88, 87.0]) -> [1, 'Ana', 90, 85, 88, 87.67]
```

````python
def pivot_row(row):
    student_id, name, *scores, old_average = row
    new_average = round(sum(scores) / len(scores), 2)
    return [student_id, name, *scores, new_average]
````

**Test cases:**

````python
print(pivot_row([1, "Ana", 90, 85, 88, 87.0]))       # expected: [1, 'Ana', 90, 85, 88, 87.67]
print(pivot_row([2, "Ben", 100, 100, 0.0]))          # expected: [2, 'Ben', 100, 100, 100.0]
print(pivot_row([3, "Cy", 70, 80, 90, 100, 0.0]))    # expected: [3, 'Cy', 70, 80, 90, 100, 85.0]
````

## Problem 7: Username Collision Resolver — `resolve_username()` (Hardest)

Accepts `first_name`, `last_name`, and a list `existing_usernames`. Builds the base username
the same way as `make_username()` from Experiment 1. If that username already exists in
`existing_usernames`, appends a number starting at 1, incrementing until the name is unique —
without using a `while` loop.

```
resolve_username("Ada", "Lovelace", ["ada.lovelace"]) -> "ada.lovelace1"
resolve_username("Ada", "Lovelace", ["ada.lovelace", "ada.lovelace1"]) -> "ada.lovelace2"
resolve_username("Alan", "Turing", []) -> "alan.turing"
```

**Hint:** Use `next()` with a generator expression over `itertools.count()`.

````python
from itertools import count

def make_username(first_name, last_name):
    clean_first = first_name.lower().replace(" ", "")
    clean_last = last_name.lower().replace(" ", "")
    return f"{clean_first}.{clean_last}"

def resolve_username(first_name, last_name, existing_usernames):
    base = make_username(first_name, last_name)
    if base not in existing_usernames:
        return base
    candidate = next(
        f"{base}{n}" for n in count(1) if f"{base}{n}" not in existing_usernames
    )
    return candidate
````

**Test cases:**

````python
print(resolve_username("Ada", "Lovelace", ["ada.lovelace"]))
# expected: ada.lovelace1

print(resolve_username("Ada", "Lovelace", ["ada.lovelace", "ada.lovelace1"]))
# expected: ada.lovelace2

print(resolve_username("Alan", "Turing", []))
# expected: alan.turing
````
